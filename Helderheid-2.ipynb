```python
import cv2

IMAGE_PATH = "bewijs.jpg"


image = cv2.imread(IMAGE_PATH)
if image is None:
    raise FileNotFoundError(f"Afbeelding niet gevonden: {IMAGE_PATH}")
```

```python
import cv2
import numpy as np

# 1. Afbeelding inlezen in grijswaarden
img = cv2.imread('bewijs.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Fout: Kan 'bewijs.jpg' niet vinden of openen.")
    exit()


img_eq = cv2.equalizeHist(img)


clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
img_clahe = clahe.apply(img)


_, thresh_otsu = cv2.threshold(
    img_clahe, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

thresh_adapt_mean = cv2.adaptiveThreshold(
    img_clahe, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 15, 8
)

thresh_adapt_gaussian = cv2.adaptiveThreshold(
    img_clahe, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 21, 10
)


cv2.imwrite('resultaat_equalized.jpg', img_eq)
cv2.imwrite('resultaat_clahe.jpg', img_clahe)
cv2.imwrite('resultaat_otsu.jpg', thresh_otsu)
cv2.imwrite('resultaat_adaptive_mean.jpg', thresh_adapt_mean)
cv2.imwrite('resultaat_adaptive_gaussian.jpg', thresh_adapt_gaussian)

print("Bewerking voltooid. Controleer de opgeslagen bestanden!")
```
