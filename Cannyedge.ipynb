```python
import cv2

IMAGE_PATH = "borden.jpg"

import numpy as np
import matplotlib.pyplot as plt 
```

```python
original = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
equ = cv2.equalizeHist(original)
res = np.hstack((original, equ))
plt.figure(figsize=(10, 5))
plt.imshow(res, cmap='gray')  
plt.title("Original vs Equalized Image")
plt.axis('off')  
plt.show()
```

```python
t_lower = 20
t_upper = 50

edge = cv2.Canny(original, t_lower, t_upper)

plt.figure(figsize=(12, 8))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(original, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edge, cmap="gray")
plt.title("Edge")
plt.axis("off")

plt.show()
```

```python
lines = cv2.HoughLinesP(
    edge,
    1,
    np.pi / 180,
    threshold=80,
    minLineLength=100,
    maxLineGap=20
)

result = cv2.cvtColor(original, cv2.COLOR_GRAY2RGB)

for i in range(len(lines)):
    x1 = lines[i][0][0]
    y1 = lines[i][0][1]
    x2 = lines[i][0][2]
    y2 = lines[i][0][3]

    cv2.line(result, (x1, y1), (x2, y2), (0, 255, 0), 2)

plt.figure(figsize=(12, 8))
plt.imshow(result)
plt.title("Hough-transformatie")
plt.axis("off")
plt.show()
```
