```python
import cv2

IMAGE_PATH = "borden.jpg"

import numpy as np
import matplotlib.pyplot as plt 
```

```python
equ = cv2.equalizeHist(IMAGE_PATH)
res = np.hstack((IMAGE_PATH, equ))
plt.figure(figsize=(10, 5))
plt.imshow(res, cmap='gray')  
plt.title("Original vs Equalized Image")
plt.axis('off')  
plt.show()
```
